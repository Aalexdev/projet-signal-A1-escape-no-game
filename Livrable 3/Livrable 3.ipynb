## Encodage

Une fois que nous avons converti notre caractère : nous pouvons l'encoder.
Chaques bits est assigné à une fréquence en particulié $f_n$ ou $n$ est le $n$ ème bit. Et $S_n(t)$ une sinusoïdale de fréquence $f_n$.
$$
S_n(t) = \sin(2 \pi f_n)
$$ 
Le signal du charactère $F(t)$ est représenté par la somme des sinusoïdales de ses bits allumé.

Ou $A_n$ représente le $n$ ème bit de caractère $A$.

$$
F_A(t) = \sum_{n=0}^8 \begin{cases}
    S_n(t) &\text{si } A_n = 1\\
    0 &\text{sinon}
\end{cases} 
$$

Dans le cas du caractère 'a' : $F_A(t) = S_1(t) + S_2(t) + S_7(t)$

Pour pouvoir le décoder facilement. Les fréquences des bits doivents suivrent une certaine logique.
Soit $f_0$ La fréquence fondamentale, la fréquence du premier bit $n=0$.

$$
f_n = n \times f_0
$$

## Modulation

Nous pouvons remarquer un problème. En effet, nous avons une plage assez limité (20-23Khz).
Si nous poseons $f_0 = 20kHz$, alors $f_1 = 40kHz$, ce qui est en dehors de notre plage.


Nous ne pouvons donc pas transmettre directement le message encodé. Nous devons donc le moduler afin de pouvoir le transmettre.

Nous avons différentes options :
 - FM : **F**requency **M**odulation
 - FSK : **F**requency **S**hift **K**eying
 - AM : **A**mplitude **M**odulation
 - ASK : **A**mplitude **S**hift **K**eying
 - PM : **P**hase **M**odulation
 - PSK : **P**hase **S**hift **K**eying

Pour savoir lequel nous devons utiliser, nous devons lister nos constraintes.
 - Un signal analogique (une function plutôt que des données binaires)
 - Une plage a disposition

Étant donné que nous voulons transmettre un signal analogique, nous devons utiliser soit FM, AM ou PM.

 - AM n'utilse qu'une seule fréquence et est simple à générer, mais est très sensibles aux interferes
 - FM utilise une plages de fréquences, est beaucoup moins sensible aux interférences mais est plus difficile à générer et démoduler.
 - PM utilise également une plage, est peu sensible aux interférances mais est plus difficile à moduler et démoduler.

Nous utiliserons ainsi la modulation FM.

$$
    s(t) = A \cos \left(2 \pi f_c t + 2 \pi k_f \int_0^tm(\tau)d\tau \right)
$$

où :
 - $A$ est l'amplitude de la proteuse
 - $f_c$ est la fréquence de la porteuse
 - $k_f$ est le facteur de modulation
 - $m(t)$ est le signal à moduler

Nous pouvons déjà ignorer $A$ car peut importe la condition, $A=1$.

Nous avons un plage $P = [P_{min}, P_{max}]$. Par exemple $P_{min} = 20000$ et $P_{max} = 23000$<br/>
Ainsi, la fréquence de la porteuse serait :
$$
    f_c = \frac {P_{min} + P_{max}} 2
$$

Le facteur de modulation serait :
$$
k_f = f_c - P_{min} = P_{max} - f_c
$$
