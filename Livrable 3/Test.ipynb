## Lecture du message

Tout d'abord, nous devons aquérir les informations que nous voulons transmettre. Ce message peut être constitué de texte. Mais étant donné que nous transmettons des octéts. Nous pouvons virtuellement transmettre toutes informations qu'un ordinateur puisse traîter.

Dans le cas d'un simple message, Nous devons itérer à travers chaques caractères du message et le traiter.

## Conversion en binaire

Notre message précedement établi a besoin d'être traduit par notre système afin d'être plus tard encoder; pour convertir le signal en binaire nous allons utiiser une un assemblage de fonction python pour traduire chaque caractère en sa valeurs du tableau ASCII. Nallons écrire ous utiliserons ensuite l'expression binaire du nombre.

Par exemple. Le caractères 'a' à comme code ASCII 97, en binaire : '01100001'.
$$
A = 01100001_2
$$


## Encodage

Une fois que nous avons converti notre caractère : nous pouvons l'encoder.
Chaques bits est assigné à une fréquence en particulié $f_n$ ou $n$ est le $n$ ème bit. Et $S_n(t)$ une sinusoïdale de fréquence $f_n$.
$$
S_n(t) = \sin(2 \pi f_n)
$$ 
Le signal du charactère $F(t)$ est représenté par la somme des sinusoïdales de ses bits allumé.

Ou $A_n$ représente le $n$ ème bit de caractère $A$.

$$
F_A(t) = \sum_{n=0}^8 \begin{cases}
    S_n(t) &\text{si } A_n = 1\\
    0 &\text{sinon}
\end{cases} 
$$

Dans le cas du caractère 'a' : $F_A(t) = S_1(t) + S_2(t) + S_7(t)$
